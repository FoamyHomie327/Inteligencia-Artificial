```python
import pandas as pd

keys = ['cve_ent', 'con', 'upm', 'n_pro_viv', 'v_sel', 'n_hog', 'h_mud', 'n_ren']

cols_sdem = keys + [
    'r_def', 'c_res',
    'ingocup',
    'sex', 'eda', 'anios_esc', 'niv_ins', 'e_con', 'n_hij',
    'hrsocup', 'rama_est1', 'pos_ocu', 'emp_ppal', 'ambito2',
    'imssissste', 'seg_soc', 'tue1', 'zona', 'ur',
    'pre_asa'
]
cols_coe2 = keys + ['p6d', 'p6c']

sdem = pd.read_csv('Datos/ENOE_SDEMT126.csv', usecols=cols_sdem, low_memory=False, encoding='latin1')
coe2 = pd.read_csv('Datos/ENOE_COE2T126.csv', usecols=cols_coe2, low_memory=False, encoding='latin1')

print("SDEM cargado:", sdem.shape)
print("COE2 cargado:", coe2.shape)

# --- Diagnóstico ANTES de filtrar nada ---
print("r_def valores:", sdem['r_def'].unique()[:10])
print("c_res valores:", sdem['c_res'].unique()[:10])

# --- Filtros ---
sdem = sdem[sdem['r_def'] == 0]
print("Después de filtrar r_def:", sdem.shape)

sdem = sdem[sdem['c_res'].isin([1, 3])]
print("Después de filtrar c_res:", sdem.shape)

sdem['eda'] = pd.to_numeric(sdem['eda'], errors='coerce')
sdem = sdem[sdem['eda'].between(12, 98)]
print("Después de filtrar eda:", sdem.shape)

# --- Merge ---
df = sdem.merge(coe2, on=keys, how='inner', suffixes=('', '_coe2'))
print("Después del merge:", df.shape)

# --- Quitar sin ingreso ---
df['ingocup'] = pd.to_numeric(df['ingocup'], errors='coerce')
df = df.dropna(subset=['ingocup'])
df = df[df['ingocup'] > 0]
print("Después de quitar sin ingreso:", df.shape)

# --- Guardar ---
df.to_csv('Datos/enoe_2026_1t_limpio.csv', index=False, encoding='utf-8')
print("Guardado como enoe_2026_1t_limpio.csv")

df.columns
df.rama_est1

print("pre_asa == 0:", (df['pre_asa'] == 0).sum())
```
